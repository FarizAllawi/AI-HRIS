```python
import pandas as pd
import os

from fastapi.params import Depends
```

```python
# LOAD CSV file
current_dir = os.getcwd()
df = pd.read_csv(current_dir+'/applicant_answers.csv')
df.head()
```

```python
# Validate Columns
required_columns = [
    'user_id',
    'applicant_id',
    'question_id',
    'answer_id',
    'job_posting_id',
    'job_posting_title',
    'job_posting_description',
    'job_posting_responsibilities',
    'job_posting_requirements',
    'job_posting_qualifications',
    'job_posting_required_skills',
    'job_posting_preferred_skills',
    'question',
    'answer',
    'weight',
    'mapped_competencies',
    'weight_version'
]

missing_columns = [col for col in required_columns if col not in df.columns]
if missing_columns:
    raise ValueError('Missing columns: {}'.format(missing_columns))

# CSV not have missing columns
print(f"✅ CSV file read successfully with {len(df)} rows.")
```

```python
# Group Job Posting data
job_posting_columns = [
    'job_posting_id',
    'job_posting_title',
    'job_posting_description',
    'job_posting_responsibilities',
    'job_posting_requirements',
    'job_posting_qualifications',
    'job_posting_required_skills',
    'job_posting_preferred_skills',
]

job_postings_list = (
    df[job_posting_columns]
    .drop_duplicates(subset=['job_posting_id'])
    .to_dict(orient='records')
)
pd.DataFrame(job_postings_list)
```

```python
# Group Job Posting Question data
job_posting_question_columns = [
    'question_id',
    'job_posting_id',
    'question',
    'weight',
    'mapped_competencies',
    'weight_version'
]

job_posting_question_list = (
    df[job_posting_question_columns]
    .drop_duplicates(subset=['question_id', 'job_posting_id', 'question'])
    .to_dict(orient='records')
)

pd.DataFrame(job_posting_question_list)
```

```python
from app.models import JobPosting

db = SessionLocal()

def create_job_posting(job_postings: list):
    for job_posting in job_postings:

        # Validate if data already exists or not
        job = db.query(JobPosting).filter(JobPosting.id == job_posting['job_posting_id']).first()
        if job:
            continue
        job_posting_data = JobPosting(
            id=job_posting['job_posting_id'],
            title=job_posting['job_posting_title'],
            description=job_posting['job_posting_description'],
            requirements=job_posting['job_posting_requirements'],
            responsibilities=job_posting['job_posting_responsibilities'],
            qualifications=job_posting['job_posting_qualifications'],
            required_skills=job_posting['job_posting_required_skills'],
            preferred_skills=job_posting['job_posting_preferred_skills'],
        )
        db.add(job_posting_data)
        db.commit()
        db.refresh(job_posting_data)
create_job_posting(job_postings_list)
```

```python
from app.models import JobPostingQuestion

db = SessionLocal()

def create_job_posting_question(questions: list):
    for question in questions:
        # Validate if data already exists or not
        exist = db.query(JobPostingQuestion).filter(JobPostingQuestion.id == question['question_id']).first()
        if exist:
            continue

        questionData = JobPostingQuestion(
            id=question['question_id'],
            job_posting_id=question['job_posting_id'],
            question=question['question'],
            weight=question['weight'],
            mapped_competencies=['mapped_competencies'],
            weight_version=question['weight_version']
        )
        db.add(questionData)
        db.commit()
        db.refresh(questionData)
create_job_posting_question(job_posting_question_list)
```
